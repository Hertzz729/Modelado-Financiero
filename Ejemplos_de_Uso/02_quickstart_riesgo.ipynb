
# Quickstart — `opciones.riesgo`

Cuatro funciones, cada una responde una pregunta distinta sobre el riesgo
de un portafolio (no de una opción individual):

- `Vol_portafolio` — ¿cuánto se mueve el portafolio, dado cómo se mueven
  y se correlacionan sus activos?
- `VaR_parametrico` — VaR asumiendo retornos normales.
- `VaR_historico` — VaR usando directamente la serie histórica de
  retornos, sin asumir ninguna distribución.
- `VaR_montecarlo` — VaR simulando trayectorias (movimiento browniano
  geométrico).


```python
from modelado_financiero.riesgo import (
    VaR_parametrico,
    VaR_historico,
    VaR_montecarlo,
    Vol_portafolio,
)

import numpy as np
```


## `Vol_portafolio`

Portafolio de 2 activos: 60% / 40%, con esta matriz de covarianzas
(varianzas en la diagonal, covarianza en la 12 y 21):


```python

pesos = [0.6, 0.4]
covarianzas = [[0.04, 0.01],
               [0.01, 0.09]]

Vol_portafolio(pesos, covarianzas)
```


## `VaR_parametrico`

Portafolio de \$1,000,000, retorno esperado diario de 0.05%, volatilidad
diaria de 1.7%, al 95% de confianza.


```python

VaR_parametrico(portafolio_valor=1_000_000, mu=0.0005, sigma=0.017, nivel_confianza=0.95)
```


## `VaR_historico`

Con una serie de 10 retornos diarios observados (en fracción, no en %),
sobre una posición de \$500,000, al 90% de confianza.


```python

retornos = np.array([0.3, -0.2, -1.1, 0.5, -0.7, 0.1, -0.3, -1.5, 0.2, -0.4]) / 100

VaR_historico(retornos, portafolio_valor=500_000, nivel_confianza=0.90)
```


## `VaR_montecarlo`

Posición de \$800,000, retorno esperado anual de 10%, volatilidad anual
de 30%, horizonte de 1 día, al 99% de confianza.


```python

VaR_montecarlo(portafolio_valor=800_000, mu=0.10, sigma=0.30, t=1/252,
                   n_simulaciones=20_000, nivel_confianza=0.99, seed=1)
```
